```python
# Columns
col_total = "Total Num Reads (Azenta)"
col_assigned_pct = "% Reads Assigned"
col_rrna_pct = "% Reads rRNA "
col_unmapped_pct = "% Unassigned: Unmapped "
col_nofeat_pct = "% Unassigned: No Features "

# Ensure numeric
df[col_total] = pd.to_numeric(df[col_total], errors="coerce")

def pct_to_fraction(series):
    s = series.astype(str).str.replace("%","", regex=False).str.strip()
    return pd.to_numeric(s, errors="coerce")/100.0

# Fractions
assigned_f = pct_to_fraction(df[col_assigned_pct]) if col_assigned_pct in df.columns else None
rrna_f = pct_to_fraction(df[col_rrna_pct]) if col_rrna_pct in df.columns else None
unmapped_f = pct_to_fraction(df[col_unmapped_pct]) if col_unmapped_pct in df.columns else None
nofeat_f = pct_to_fraction(df[col_nofeat_pct]) if col_nofeat_pct in df.columns else None

# Recalculate existing count columns (only if present); do not insert/reorder
def recalc(target_col, frac_series):
    if target_col in df.columns and frac_series is not None:
        df[target_col] = (df[col_total] * frac_series).round().astype("Int64")

recalc("Reads Assigned", assigned_f)
recalc("Reads rRNA", rrna_f)
recalc("Reads Unassigned: Unmapped", unmapped_f)
recalc("Reads Unassigned: No Features", nofeat_f)
```

```python
cols_to_describe = [
    "Reads Assigned",
    "Reads rRNA",
    "Reads Unassigned: Unmapped",
    "Reads Unassigned: No Features",
    "RQN (Azenta)"
]
```

```python
import numpy as np
import matplotlib.ticker as ticker
import matplotlib.pyplot as plt
from pathlib import Path

# Create histograms with x-axis in millions and bins of 1 million reads
fig_paths = []
for c in cols_to_describe:
    if c not in df.columns:
        continue
    s = pd.to_numeric(df[c], errors="coerce").dropna()
    if s.empty:
        continue

    # Convert to millions for readability
    s_millions = s / 1e6
    bins = np.arange(0, np.ceil(s_millions.max()) + 1, 1)  # 1M bin width

    plt.figure(figsize=(6, 4))
    plt.hist(s_millions, bins=bins, edgecolor='black')
    plt.xlabel(f"{c} (million reads)")
    plt.ylabel("Number of samples")
    plt.title(f"Distribution of {c}")
    plt.grid(axis='y', alpha=0.3)

    # Format x-axis to show e.g., 1, 2, 3 instead of scientific notation
    plt.gca().xaxis.set_major_locator(ticker.MaxNLocator(integer=True))
    plt.gca().xaxis.set_minor_locator(ticker.AutoMinorLocator())

    out = Path(f"{c.replace(' ', '_').replace(':','_').replace('(','').replace(')','')}_hist_millions.png")
    plt.tight_layout()
    plt.savefig(out, dpi=150)
    plt.close()
    fig_paths.append(out.as_posix())

fig_paths
```

```python
# Convert to numeric where needed
num_cols = [
    "Total Num Reads (Azenta)",
    "Reads Assigned",
    "Reads rRNA",
    "Reads Unassigned: Unmapped",
    "Reads Unassigned: No Features",
    "RQN (Azenta)",
    "Mean Quality Score (Azenta)",
]

for col in num_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Function for median + IQR
def median_iqr(series):
    s = series.dropna()
    if s.empty:
        return np.nan, np.nan, np.nan
    q1, med, q3 = np.percentile(s, [25, 50, 75])
    return med, q1, q3

# Summary values
n_samples = len(df)
n_strains = df["Strain"].nunique() if "Strain" in df.columns else np.nan
n_conditions = df["Growth Medium"].nunique() if "Growth Medium" in df.columns else np.nan

total_med, total_q1, total_q3 = median_iqr(df["Total Num Reads (Azenta)"])
assigned_med, assigned_q1, assigned_q3 = median_iqr(df["Reads Assigned"])
rrna_med, rrna_q1, rrna_q3 = median_iqr(df["Reads rRNA"])
rqn_med, rqn_q1, rqn_q3 = median_iqr(df["RQN (Azenta)"])
read_qual_mean = df["Mean Quality Score (Azenta)"].mean()
read_qual_sd = df["Mean Quality Score (Azenta)"].std()

# Create manuscript-ready text
text = f"""
The RNA-Seq compendium comprised {n_samples:,} libraries from {n_strains} actinomycete strains grown in {n_conditions} media. 
Libraries yielded a median of {total_med/1e6:.1f} million reads (IQR: {total_q1/1e6:.1f}–{total_q3/1e6:.1f} million), 
of which a median of {assigned_med/1e6:.1f} million (IQR: {assigned_q1/1e6:.1f}–{assigned_q3/1e6:.1f} million) were assigned to features, 
and {rrna_med/1e6:.1f} million (IQR: {rrna_q1/1e6:.1f}–{rrna_q3/1e6:.1f} million) corresponded to rRNA reads. 
RNA integrity was high (median RQN = {rqn_med:.1f}, IQR: {rqn_q1:.1f}–{rqn_q3:.1f}), and mean read quality was {read_qual_mean:.2f} ± {read_qual_sd:.2f}. 
Although most strain–condition combinations were represented by a single library, mapping and expression distributions were consistent across the dataset, 
indicating high technical reproducibility.
"""

text
```

```python
import pandas as pd

# Convert reads assigned to numeric
df["Reads Assigned"] = pd.to_numeric(df["Reads Assigned"], errors="coerce")

# Keep samples with >= 1M assigned reads
filtered = df[df["Reads Assigned"] >= 1e6].copy()

# The 8 required media
required_media = ["DNPM", "ISP2", "MA", "soyM", "TSB50", "SMM-glucose", "SMM-glycerol", "SMM-maltose"]

# --------------------------------------------------
# 1) Keep only samples where strain-medium has >1 replicate
# --------------------------------------------------
rep_counts = filtered.groupby(["Strain", "Growth Medium"])["Sample_name"].transform("size")
df_more_than_one_rep = filtered[rep_counts > 1].copy()

print("Samples in strain-medium groups with >1 replicate:")
print(df_more_than_one_rep[["Sample_name", "Strain", "Growth Medium", "Replicate"]].head())

# --------------------------------------------------
# 2) Count replicates per strain and medium
# --------------------------------------------------
rep_table = (
    filtered.groupby(["Strain", "Growth Medium"])
    .size()
    .unstack(fill_value=0)
)

# Ensure all 8 media are included
rep_table = rep_table.reindex(columns=required_media, fill_value=0)

# --------------------------------------------------
# 3) Find strains with exactly 2 replicates in all 8 media
# --------------------------------------------------
strains_with_2reps_all8 = rep_table[(rep_table == 2).all(axis=1)].index.tolist()

print("\nStrains with exactly 2 replicates in all 8 media:")
print(strains_with_2reps_all8)
print(f"Number of strains: {len(strains_with_2reps_all8)}")

# --------------------------------------------------
# 4) Subset dataframe to only those strains
# --------------------------------------------------
df_strains_2reps_all8 = filtered[
    filtered["Strain"].isin(strains_with_2reps_all8)
].copy()

# --------------------------------------------------
# 5) Final subset: samples from qualifying strains
#    where the strain-medium group has >1 replicate
# --------------------------------------------------
df_final = df_more_than_one_rep[
    df_more_than_one_rep["Strain"].isin(strains_with_2reps_all8)
].copy()

print("\nFinal dataframe shape:")
print(df_final.shape)

# Optional: show replicate count table for qualifying strains
print("\nReplicate table for qualifying strains:")
print(rep_table.loc[strains_with_2reps_all8])
```
