```python
# wide -> long (one row per Orthogroup x Strain)
long = og.melt(
    id_vars="Orthogroup",
    var_name="Strain",
    value_name="Locus_list"
)

# drop empty cells
long = long.dropna(subset=["Locus_list"])
long["Locus_list"] = long["Locus_list"].astype(str).str.strip()
long = long[long["Locus_list"] != ""]

# split comma-separated locus IDs -> one locus per row
long["Geneid"] = long["Locus_list"].str.split(r",\s*")
long = long.explode("Geneid", ignore_index=True)

# cleanup
long["Geneid"] = long["Geneid"].str.strip()
long = long.drop(columns=["Locus_list"])

long.head()
```

```python
import os
import glob
import pandas as pd

# --- paths ---
base_dir = "/Users/annasve/Desktop/article_data/input/gene_expression_metadata"

# --- long df must exist already ---
# Ensure correct column name:
if "Locus_ID" in long.columns and "Geneid" not in long.columns:
    long = long.rename(columns={"Locus_ID": "Geneid"})

required = {"Orthogroup", "Strain", "Geneid"}
missing = required - set(long.columns)
if missing:
    raise ValueError(f"`long` is missing columns: {missing}")

# (Optional) if long has duplicates per (Strain, Geneid), keep first to avoid row multiplication
long_keyed = long.drop_duplicates(subset=["Strain", "Geneid"]).copy()

def read_table(path: str) -> tuple[pd.DataFrame, str]:
    """Read csv/tsv, return df and separator used."""
    ext = os.path.splitext(path)[1].lower()
    if ext in [".tsv", ".txt"]:
        return pd.read_csv(path, sep="\t", low_memory=False), "\t"
    else:
        # default to CSV
        return pd.read_csv(path, sep=",", low_memory=False), ","

def write_atomic(df: pd.DataFrame, path: str, sep: str):
    """Write to temp then replace original (safer than direct overwrite)."""
    tmp_path = path + ".tmp"
    df.to_csv(tmp_path, sep=sep, index=False)
    os.replace(tmp_path, path)

# Iterate strain folders
strain_dirs = [d for d in glob.glob(os.path.join(base_dir, "*")) if os.path.isdir(d)]

for strain_dir in strain_dirs:
    strain = os.path.basename(strain_dir)

    # Find files to process (adjust patterns if you only want certain files)
    files = []
    files += glob.glob(os.path.join(strain_dir, "*.csv"))
    files += glob.glob(os.path.join(strain_dir, "*.tsv"))
    files += glob.glob(os.path.join(strain_dir, "*.txt"))

    if not files:
        continue

    # mapping for this strain
    map_strain = long_keyed[long_keyed["Strain"] == strain][["Geneid", "Orthogroup"]]

    for fp in files:
        df, sep = read_table(fp)

        # 1) remove Orthogroup if present
        if "Orthogroup" in df.columns:
            df = df.drop(columns=["Orthogroup"])

        # 2) merge based on Geneid
        if "Geneid" not in df.columns:
            # If some files still use Locus_ID, you could auto-handle here:
            if "Locus_ID" in df.columns:
                df = df.rename(columns={"Locus_ID": "Geneid"})
            else:
                # skip files that don't have the join key
                print(f"SKIP (no Geneid): {fp}")
                continue

        merged = df.merge(map_strain, on="Geneid", how="left")

        # 3) overwrite same file
        write_atomic(merged, fp, sep=sep)

print("Done.")
```

```python
from pathlib import Path
import pandas as pd
import os

# --------------------------------------------------
# PATHS
# --------------------------------------------------
base_dir = Path("/Users/annasve/Desktop/article_data/expression_module")

# --------------------------------------------------
# long dataframe must already exist
# --------------------------------------------------
if "Locus_ID" in long.columns and "Geneid" not in long.columns:
    long = long.rename(columns={"Locus_ID": "Geneid"})

required = {"Orthogroup", "Strain", "Geneid"}
missing = required - set(long.columns)
if missing:
    raise ValueError(f"`long` is missing columns: {missing}")

# avoid duplicate merge expansion
long_keyed = long.drop_duplicates(subset=["Strain", "Geneid"]).copy()

# --------------------------------------------------
# HELPERS
# --------------------------------------------------
def write_excel_atomic(df: pd.DataFrame, path: Path):
    """Write to temp file, then replace original."""
    tmp_path = path.with_suffix(path.suffix + ".tmp.xlsx")
    df.to_excel(tmp_path, index=False, engine="openpyxl")
    os.replace(tmp_path, path)

# --------------------------------------------------
# FIND FILES
# --------------------------------------------------
files = list(base_dir.rglob("*bgc_borders.xlsx"))
print(f"Found {len(files)} files")

error_log = []

for fp in files:
    strain = fp.parent.name
    print(f"Processing: {fp}")

    try:
        df = pd.read_excel(fp, engine="openpyxl")
    except Exception as e:
        msg = f"{fp} -> READ ERROR: {e}"
        print(msg)
        error_log.append(msg)
        continue

    # remove old Orthogroup if present
    if "Orthogroup" in df.columns:
        df = df.drop(columns=["Orthogroup"])

    # make sure join key exists
    if "Geneid" not in df.columns:
        if "Locus_ID" in df.columns:
            df = df.rename(columns={"Locus_ID": "Geneid"})
        else:
            msg = f"{fp} -> SKIP: no Geneid or Locus_ID column"
            print(msg)
            error_log.append(msg)
            continue

    # mapping for this strain
    map_strain = long_keyed[long_keyed["Strain"] == strain][["Geneid", "Orthogroup"]]

    # merge Orthogroup in
    merged = df.merge(map_strain, on="Geneid", how="left")

    # overwrite same file
    try:
        write_excel_atomic(merged, fp)
        print(f"Replaced: {fp}")
    except Exception as e:
        msg = f"{fp} -> WRITE ERROR: {e}"
        print(msg)
        error_log.append(msg)

# --------------------------------------------------
# SAVE ERRORS
# --------------------------------------------------
if error_log:
    log_path = base_dir / "orthogroup_merge_errors.txt"
    with open(log_path, "w") as f:
        for line in error_log:
            f.write(line + "\n")
    print(f"Error log saved to: {log_path}")

print("Done.")
```

```python
import os
import glob
import pandas as pd

base_dir = "/Users/annasve/Desktop/article_data/input/gene_expression_metadata"

dfs = []

for strain_dir in glob.glob(os.path.join(base_dir, "*")):
    if not os.path.isdir(strain_dir):
        continue

    strain = os.path.basename(strain_dir)

    files = glob.glob(os.path.join(strain_dir, "*.csv"))  # adjust if needed

    if not files:
        continue

    # mapping for this strain
    map_strain = (
        long[long["Strain"] == strain][["Geneid", "Orthogroup"]]
        .drop_duplicates()
    )

    for fp in files:
        df = pd.read_csv(fp, low_memory=False)

        if "Geneid" not in df.columns:
            continue

        if "Orthogroup" in df.columns:
            df = df.drop(columns=["Orthogroup"])

        merged = df.merge(map_strain, on="Geneid", how="left")

        merged["Strain"] = strain
        merged["Source_file"] = os.path.basename(fp)

        dfs.append(merged)

all_merged = pd.concat(dfs, ignore_index=True)
```

```python
og_top_product = (
    all_merged
    .dropna(subset=["Product"])
    .groupby("Orthogroup")["Product"]
    .agg(lambda s: s.value_counts().idxmax())
    .reset_index(name="top_product")
)
```

```python
def og_product_stats(df):
    # only rows with an orthogroup
    d = df.dropna(subset=["Orthogroup"]).copy()

    # total genes per orthogroup
    total = (
        d.groupby("Orthogroup")
        .size()
        .rename("n_total_genes")
    )

    # genes with a product annotation
    annotated = (
        d.dropna(subset=["Product"])
        .groupby("Orthogroup")
        .size()
        .rename("n_annotated")
    )

    # most common product + how often it appears
    top = (
        d.dropna(subset=["Product"])
        .groupby("Orthogroup")["Product"]
        .value_counts()
        .groupby(level=0)
        .agg(
            top_product=lambda s: s.index[0][1],
            top_count=lambda s: s.iloc[0]
        )
    )

    # combine
    out = (
        pd.concat([total, annotated, top], axis=1)
        .fillna(0)
        .reset_index()
    )

    # derived metrics
    out["n_annotated"] = out["n_annotated"].astype(int)
    out["n_total_genes"] = out["n_total_genes"].astype(int)
    out["top_count"] = out["top_count"].astype(int)

    out["n_not_annotated"] = out["n_total_genes"] - out["n_annotated"]
    out["fraction_annotated"] = out["n_annotated"] / out["n_total_genes"]
    out["fraction_top"] = out["top_count"] / out["n_annotated"].replace(0, pd.NA)

    return out

og_summary = og_product_stats(all_merged)
og_summary.head()
```
